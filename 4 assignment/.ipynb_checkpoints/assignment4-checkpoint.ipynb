### Summary

Statistics: there are in total 5 AI results that match with manual checking, i.e 25% rate. 

The post-game interviews are really specific type of emotions expression because it usually relates to cases when managers are prone to conceal and even withhold their real sentiment, speaking in a highly veiled manner, just to keep the intensity (if it has) inside the club without unnecessary publicity. Obviously, context matters in many cases, but to be honest it is not easy deal for people who non-dived into football realm to fully comprehend the idea that managers try to come with and convey after the games. 

There is always duplicity in a neutral manner. What I mean is that the manager seeks not to over-praise the team even if it showed a great perfomance. There are really seldom cases when a coach praises his team without at least small comments or reproaches. It is kind of a part of managers' strategy to keep the team in stability. Analogically, a coach rather mitigates his resentment (and even outrage) saying something like 'we could perform better today', 'I am not happy, but we did what we were able to do today'. Though, in reality he is going to make a harsh discussion with a team because of an awful performance. 

All in all, if managers are too emotional, it may rather affect negatively the atmosphere inside the club because of media or fans pressure. The coach can say that everything is terrible and it will be exaggerated by the media, at the same time the coach says that it couldn't be better, everythig is so ideal that eventually create inflated expectations for the team and as a result additional pressure. 

As we can see, AI is likely not able to catch these word tricks. In my case, managers mostly used as many supporting and positive collocations as negative and judjment phrases at the same time. So, AI couldn't find the bottom line among these word mess and just set a neutral estimation for the most part of the observations. We can observe that more than half of rows contain zero while the manual check demonstrates absolutely different outcome.
