# Browsing datasets

```python
import pandas as pd

# Задаём путь к папке
folder_path = r'C:/Users/Пользователь/Desktop/MA1y/Data_Analysis_AI/DAT_AI/3 assignment_(group)'

# Список файлов
file_names = [
    'amenities.csv',
    'cities_modified.csv',
    'city_tourism_modified.csv',
    'economic_indicators.csv',
    'hotel_amenities_modified.csv',
    'hotels_modified.csv',
    'monthly_occupancy_modified.csv',
    'reviews_modified.csv'
]

dfs={}
for file_name in file_names:
    file_path = os.path.join(folder_path, file_name)
    
    # Check if the file exists before trying to open it
    if os.path.exists(file_path):
        try:
            # Create a variable name based on the file name (without the '.csv' extension)
            df_name = file_name.replace('.csv', '')  # Remove the '.csv' part
            
            # Load the CSV into the DataFrame
            dfs[df_name] = pd.read_csv(file_path)
            
            # Print the first 5 rows of the DataFrame
            print(f"\nSuccessfully loaded: {df_name}")
            print(dfs[df_name].head())  # Show the first 5 rows
        except Exception as e:
            print(f"Error reading {file_name}: {e}")
    else:
        print(f"File not found: {file_name}")
```

```python
# Удаляем строки с пропущенными значениями для каждого DataFrame в словаре
for df_name, df in dfs.items():
    # Удаляем строки с пропущенными значениями
    dfs[df_name] = df.dropna()
    
    # Выводим первые 5 строк обновленного DataFrame
    print(f"\nUpdated {df_name} (after removing missing values):")
    print(dfs[df_name].head())
```

```python
# Выводим все заголовки (имена столбцов) для каждого DataFrame в словаре dfs
for df_name, df in dfs.items():
    print(f"\nColumns in {df_name}:")
    print(df.columns.tolist())  # Выводим список всех столбцов
```

# Join city and hotel

```python
# Perform the four different types of joins
# 1. Inner Join
city_hotel_1 = pd.merge(dfs['cities_modified'], dfs['hotels_modified'], on='city', how='inner')

# 2. Left Join
city_hotel_2 = pd.merge(dfs['cities_modified'], dfs['hotels_modified'], on='city', how='left')

# 3. Right Join
city_hotel_3 = pd.merge(dfs['cities_modified'], dfs['hotels_modified'], on='city', how='right')

# 4. Outer Join
city_hotel_4 = pd.merge(dfs['cities_modified'], dfs['hotels_modified'], on='city', how='outer')
```

# Hotel and Occupancy

```python
# Perform the four different types of joins with the 'monthly_occupancy_modified' dataset
# 1. Inner Join with city_hotel_1
hotel_occupancy_1 = pd.merge(dfs['hotels_modified'], dfs['monthly_occupancy_modified'], on='hotel_id', how='inner')

# 2. Left Join with city_hotel_2
hotel_occupancy_2 = pd.merge(dfs['hotels_modified'], dfs['monthly_occupancy_modified'], on='hotel_id', how='left')

# 3. Right Join with city_hotel_3
hotel_occupancy_3 = pd.merge(dfs['hotels_modified'], dfs['monthly_occupancy_modified'], on='hotel_id', how='right')

# 4. Outer Join with city_hotel_4
hotel_occupancy_4 = pd.merge(dfs['hotels_modified'], dfs['monthly_occupancy_modified'], on='hotel_id', how='outer')

```

## AVG occup rate by hotel

```python
# Group by hotel_id, year, and month, and calculate the mean of 'occupancy_rate'
average_occupancy = dfs['monthly_occupancy_modified'].groupby(['hotel_id', 'year', 'month'], as_index=False)['occupancy_rate'].mean()

# Print the first 5 rows of the resulting DataFrame
print("\nAverage Occupancy Rate for each hotel for each year and month:")
print(average_occupancy.head())
```

```python
# Perform a left join between the average_occupancy dataset and the hotels_modified dataset
hotel_occupancy_with_details = pd.merge(average_occupancy, dfs['hotels_modified'], on='hotel_id', how='left')

# Print the first 5 rows of the resulting DataFrame
print("\nHotel Occupancy with Details (Left Join):")
hotel_occupancy_with_details
```

```python
# Merge monthly occupancy with hotels and cities
occupancy_by_city_month = pd.merge(dfs['monthly_occupancy_modified'], dfs['hotels_modified'], on='hotel_id', how='left')
occupancy_by_city_month = pd.merge(occupancy_by_city_month, dfs['cities_modified'], on='city', how='left')

# Group by city and month, and calculate the mean of occupancy_rate
occupancy_by_city_month_avg = occupancy_by_city_month.groupby(['city', 'month'], as_index=False)['occupancy_rate'].mean()

# Show result
print("\nOccupancy Rate by City and Month:")
occupancy_by_city_month_avg
```

```python
# Merge hotel data with occupancy data
star_occupancy = pd.merge(dfs['hotels_modified'], dfs['monthly_occupancy_modified'], on='hotel_id', how='left')

# Group by star_rating and calculate the mean of occupancy_rate
star_occupancy_avg = star_occupancy.groupby('star_rating', as_index=False)['occupancy_rate'].mean()

# Show result
print("\nOccupancy Rate by Hotel Star Rating:")
star_occupancy_avg
```
