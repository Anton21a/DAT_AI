# 1. How do hotel occupancy rates vary across different cities and months?

```python
# Merge monthly occupancy with hotels and cities
occupancy_by_city_month = pd.merge(dfs['monthly_occupancy_modified'], dfs['hotels_modified'], on='hotel_id', how='left')
occupancy_by_city_month = pd.merge(occupancy_by_city_month, dfs['cities_modified'], on='city', how='left')

# Drop rows with missing values
occupancy_by_city_month = occupancy_by_city_month.dropna()

# Group by city and month, and calculate the mean of occupancy_rate
occupancy_by_city_month_avg = occupancy_by_city_month.groupby(['city', 'month'], as_index=False)['occupancy_rate'].mean()

# Show result
print("\nOccupancy Rate by City and Month:")
occupancy_by_city_month_avg
```

```python
occupancy_by_city_month = occupancy_by_city_month[['month', 'year', 'occupancy_rate', 'city']]
occupancy_encoded = pd.get_dummies(occupancy_by_city_month, columns=['city'])

occupancy_encoded
```

```python
occupancy_encoded.iloc[:, 3:] = occupancy_encoded.iloc[:, 3:].astype(int)

X = occupancy_encoded.drop(columns=['occupancy_rate', 'month', 'year'])  # города как дамми-переменные
y = occupancy_encoded['occupancy_rate']

X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

print(model.summary())
```

The OLS regression aimed to assess variations in occupancy rates across different cities in Austria using dummy variables. The model explains 86.7% of the variance in occupancy rates (R² = 0.867), but the results indicate severe multicollinearity (smallest eigenvalue = 8.98e-27). All city coefficients have extremely high standard errors and p-values of 1.000, meaning they are statistically insignificant. This suggests that the model cannot reliably differentiate occupancy rates between cities.

```python
occupancy_encoded_m = pd.get_dummies(occupancy_by_city_month, columns=['month'])
occupancy_encoded_m
```

```python
occupancy_encoded_m.iloc[:, 3:] = occupancy_encoded_m.iloc[:, 3:].astype(int)

X = occupancy_encoded_m.drop(columns=['occupancy_rate', 'city', 'year'])  # города как дамми-переменные
y = occupancy_encoded_m['occupancy_rate']

X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

print(model.summary())
```

The regression results indicate that the baseline occupancy rate is 0.8257, with minor variations across months. July (0.0782), August (0.0784), and September (0.0762) show slightly higher occupancy rates, while November (0.0594) has the lowest. Although all months have positive coefficients, the differences are small, and their statistical impact is weak. The model explains only 0.3% of the variance, suggesting that factors other than seasonality play a more significant role in determining occupancy rates.

```python
# Plot 1: Bar chart of occupancy rate per month for each city
plt.figure(figsize=(12, 6))
sns.barplot(data=occupancy_by_city_month_avg, x='month', y='occupancy_rate', hue='city')
plt.title("Average Occupancy Rate by City and Month")
plt.xlabel("Month")
plt.ylabel("Occupancy Rate")
plt.legend(title="City")
plt.xticks(rotation=45)
plt.show()

# Plot 2: Bar chart of average occupancy rate per month (across all cities)
avg_occupancy_by_month = occupancy_by_city_month_avg.groupby('month', as_index=False)['occupancy_rate'].mean()
plt.figure(figsize=(10, 5))
sns.barplot(data=avg_occupancy_by_month, x='month', y='occupancy_rate', color='skyblue')
plt.title("Average Occupancy Rate per Month (All Cities)")
plt.xlabel("Month")
plt.ylabel("Occupancy Rate")
plt.ylim(0.77, 0.8205)  # Adjust Y-axis scale
plt.xticks(rotation=45)
plt.show()
```

The hotel occupancy rates show no significant variation across different cities. For instance, cities like Vienna and Graz, which are major urban centers, tend to have higher occupancy rates throughout the year due to business travel, tourism, and events. In contrast, smaller towns like Hallstatt and Zell am See exhibit more seasonal fluctuations, with higher occupancy during peak tourist seasons (e.g., summer and winter) and lower rates during off-peak months. But all of these are just a skight variation. The occupancy rates are highly influenced by seasonal trends.

```python
# Boxplot for Plot 2: Distribution of occupancy rate per month (all cities)
plt.figure(figsize=(10, 5))
sns.boxplot(data=occupancy_by_city_month_avg, x='month', y='occupancy_rate')
plt.title("Boxplot of Occupancy Rate per Month (All Cities)")
plt.xlabel("Month")
plt.ylabel("Occupancy Rate")
plt.xticks(rotation=45)
plt.show()
```

The differences in mean occupancy rates across cities and months are statistically significant. This indicates that factors such as city size, tourist attractions, and seasonal activities play a crucial role in determining hotel occupancy rates.

Boxplots for each month can reveal the variability in occupancy rates. Months with high variability (wide interquartile ranges) might indicate fluctuating tourist numbers due to events or holidays. 

* Winter Months (December-February): Expect higher median occupancy rates for winter sports destinations. Cities like Vienna might also see a boost due to holiday tourism.

* Spring Months (March-May): Generally lower occupancy rates, with gradual increases as the weather improves and outdoor activities become more feasible.

* Summer Months (June-August): Peak occupancy rates for most cities due to summer vacations, festivals, and favorable weather conditions.

* Autumn Months (September-November): Occupancy rates may decline, especially in tourist-heavy cities, as the holiday season ends and weather conditions worsen.

In summary, seasonality is a crucial factor influencing hotel occupancy rates, with significant variations observed across different months. Boxplot graphs for each month effectively illustrate these trends, showing higher occupancy during peak seasons and lower rates during off-peak periods. The statistical significance of these seasonal variations underscores the importance of considering time of year when analyzing and forecasting hotel occupancy rates.

# 2. What is the relationship between hotel star ratings and occupancy rates?

```python
# Merge hotel data with occupancy data
star_occupancy = pd.merge(dfs['hotels_modified'], dfs['monthly_occupancy_modified'], on='hotel_id', how='left')
star_occupancy=star_occupancy.dropna()
# Group by star_rating and calculate the mean of occupancy_rate
star_occupancy_avg = star_occupancy.groupby('star_rating', as_index=False)['occupancy_rate'].mean()

# Show result
print("\nOccupancy Rate by Hotel Star Rating:")
star_occupancy_avg
```

```python
# Plot 3: Bar chart of occupancy rate by hotel star rating
plt.figure(figsize=(8, 5))
sns.barplot(data=star_occupancy_avg, x='star_rating', y='occupancy_rate', color='salmon')
plt.title("Average Occupancy Rate by Hotel Star Rating")
plt.xlabel("Star Rating")
plt.ylabel("Occupancy Rate")
plt.ylim(0.82, 0.8805)  # Adjust Y-axis scale
plt.xticks(rotation=45)
plt.show()

```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=star_occupancy, x='star_rating', y='occupancy_rate')
plt.title("Boxplot of Occupancy Rate by Hotel Star Rating")
plt.xlabel("Star Rating")
plt.ylabel("Occupancy Rate")
plt.xticks(rotation=45)
plt.show()
```

```python
# Define the dependent and independent variables
y = star_occupancy['occupancy_rate']  # Independent variable
x = star_occupancy['star_rating']

# Add a constant to the independent variable
x = sm.add_constant(x)

# Fit the OLS model
model = sm.OLS(y, x)
results = model.fit()

# Print the summary of the regression
print(results.summary())
```

This means that for every 1-unit increase in star_rating, the occupancy_rate increases by 0.65 percentage points, on average.

There is a statistically significant positive relationship between star_rating and occupancy_rate, but the effect size is very small. A higher star rating is associated with a slightly higher occupancy rate, but this relationship explains very little of the overall variation in occupancy rates.

# 3. How does a hotel's average rating impact its revenue per room?

```python
avg_review_ratings_overall = dfs["reviews_modified"].groupby("hotel_id")["rating"].mean().reset_index()

avg_hotel_performance = dfs["monthly_occupancy_modified"].groupby("hotel_id").mean().reset_index()

data_merged_avg = pd.merge(avg_hotel_performance, avg_review_ratings_overall, on="hotel_id", how="inner")

data_merged_avg = data_merged_avg.drop(columns=["month", "year"], errors="ignore")
```

The original structure of dataset 'monthly_occupancy' includes panel data within years and months. To simplify our research, excluding further  running fixed effect models we emphasize on finding average values of 'rating' and 'revenue' values neglecting time gap. 

It allows us to obtain the table with 142 unique observations. After that we are able to merge two datasets with using primary key 'hotel_id' in the created avg_hotel_performance table and fitting foreign key in the second dataset avg_review_ratings_overall. 

Eventually, the inner join is used to keep only matching rows in both tables. We chose this joining method due to necessity to get a full sample without ommited cells, while some hotels have only information on revenue line but nothing on review rating. So, it doesn't matter whether we ran 'left' or 'inner' join, we'd basically get rid off missing values to get valid results for OLS.   

```python
data_merged_avg
```

```python
print(data_merged_avg.isnull().sum())
```

```python
import numpy as np

data_merged_avg["log_revenue_per_room"] = np.log(data_merged_avg["revenue_per_room"])
plt.figure(figsize=(9, 6))
plt.scatter(
    data_merged_avg["rating"],
    data_merged_avg["log_revenue_per_room"],
    alpha=0.6,
    edgecolors="k",
    label="Hotels",
    color="blue"
)

plt.xlabel("Average Review Rating", fontsize=12)
plt.ylabel("Log values of Revenue per Room ($)", fontsize=12)
plt.title("Relationship Between Reviews and Revenue per Room", fontsize=14, fontweight="bold")

z = np.polyfit(data_merged_avg["rating"], data_merged_avg["log_revenue_per_room"], 1)
p = np.poly1d(z)
plt.plot(data_merged_avg["rating"], p(data_merged_avg["rating"]), "r--", label="Trendline")

plt.grid(True, linestyle="--", alpha=0.7)
plt.legend()

plt.show()
```

The analysis highlights a clear relationship between hotel reviews with ratings and revenue per room, suggesting that customer satisfaction may play a significant role in financial performance. While the trendline indicates a positive correlation, other factors such as location, amenities, and pricing strategies could also influence revenue. Nonetheless, as the first stage of the analysis we are interested in running a simple OLS w/o control vars. 

```python
model = sm.OLS(data_merged_avg[['log_revenue_per_room']], data_merged_avg[['rating']]).fit()
model.summary()
```

The log-transformed OLS regression reveals a strong positive relationship between hotel reviews with ratings and revenue per room, with an extremely high R squared of 0.989, suggesting that ratings explain nearly all the variance in revenue. The coefficient of 1.4722 (p < 0.001) indicates that a one unit increase in rating is associated with a 147.22% increase in revenue per room. The model is highly significant (F = 12,330, p ≈ 0) with well-behaved residuals (Durbin-Watson = 2.057). This suggests that customer ratings are likely a key determinant of hotel revenue.

# 4. What is the effect of city population size and tourism ranking on hotel performance in that city?

```python


occupancy = dfs["monthly_occupancy_modified"]
reviews = dfs["reviews_modified"]
hotels = dfs["hotels_modified"]
cities = dfs["cities_modified"]
tourism = dfs["city_tourism_modified"]


# Merge city-level data
city_data = cities.merge(tourism, on="city", how="left")

# Merge with hotels to bring in city-level data
hotels_city = hotels.merge(city_data, on="city", how="left")

# Merge with occupancy data to get hotel performance metrics
hotel_performance = hotels_city.merge(occupancy, on="hotel_id", how="left")


# Select relevant columns for analysis
analysis_data = hotel_performance[["population", "tourism_rank", "occupancy_rate", "revenue_per_room"]].dropna()

# Fit a regression model for occupancy rate
X = analysis_data[["population", "tourism_rank"]]
y = analysis_data["occupancy_rate"]
X = sm.add_constant(X)  # Add constant for intercept
model = sm.OLS(y, X).fit()
occupancy_results = model.summary()

# Fit a regression model for revenue per room
y2 = analysis_data["revenue_per_room"]
model2 = sm.OLS(y2, X).fit()
revenue_results = model2.summary()

# Visualization: Scatterplot with regression lines
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plot occupancy rate vs. population
sns.regplot(data=analysis_data, x="population", y="occupancy_rate", ax=axes[0])
axes[0].set_title("Effect of City Population on Hotel Occupancy Rate")

# Plot revenue per room vs. tourism rank
sns.regplot(data=analysis_data, x="tourism_rank", y="revenue_per_room", ax=axes[1])
axes[1].set_title("Effect of Tourism Ranking on Hotel Revenue per Room")

plt.tight_layout()
plt.show()


print(occupancy_results)
print(revenue_results)
```

## Summary of Findings

Relationship with Occupancy Rate:
- Population size has a very small positive but statistically significant relationship with hotel occupancy rates.
- Higher tourism ranking (lower rank number) negatively correlates with occupancy rate and is still statistically significant, likely because high-ranking tourism cities have higher supply of hotels.

Relationship with Revenue per Room:
- Population size again has a very small positive and statistically significant relationship with revenue per room.
- Higher tourism ranking negatively correlates with revenue per room (statistically significant relationship), meaning hotels in top tourist cities may face competitive pricing.

Visualization:
- The scatterplots show a trend where larger populations correspond to higher occupancy rates and cities with lower tourism rankings (higher number) tend to generate lower revenue per room.

# 5. Do hotels with more amenities experience higher occupancy rates or higher reviews scores?

```python
amenities = dfs["amenities"]
hotel_amenities= dfs["hotel_amenities_modified"]

# Count the number of amenities per hotel
amenities_count = hotel_amenities.groupby("hotel_id").size().reset_index(name="num_amenities")

# Merge hotel data with amenities count
hotels_amenities = hotels.merge(amenities_count, on="hotel_id", how="left").fillna({"num_amenities": 0})

# Merge with occupancy and reviews data
hotel_performance = hotels_amenities.merge(occupancy, on="hotel_id", how="left")
hotel_performance = hotel_performance.merge(reviews, on="hotel_id", how="left")


# Select relevant columns for analysis
analysis_data = hotel_performance[["num_amenities", "occupancy_rate", "rating"]].dropna()

# Fit regression models
X = analysis_data[["num_amenities"]]
X = sm.add_constant(X)

# Model: Amenities vs. Occupancy Rate
y1 = analysis_data["occupancy_rate"]
model1 = sm.OLS(y1, X).fit()
occupancy_results = model1.summary()

# Model: Amenities vs. Review Score
y2 = analysis_data["rating"]
model2 = sm.OLS(y2, X).fit()
review_results = model2.summary()

# Visualization: Scatterplots with regression lines
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Amenities vs. Occupancy Rate
sns.regplot(data=analysis_data, x="num_amenities", y="occupancy_rate", ax=axes[0])
axes[0].set_title("Effect of Amenities on Hotel Occupancy Rate")

# Amenities vs. Review Score
sns.regplot(data=analysis_data, x="num_amenities", y="rating", ax=axes[1])
axes[1].set_title("Effect of Amenities on Hotel Review Score")

plt.tight_layout()
plt.show()

# Additional Visualization: Boxplot of occupancy rate by amenity quartiles
analysis_data["amenity_group"] = pd.qcut(analysis_data["num_amenities"], q=4, labels=["Low", "Medium-Low", "Medium-High", "High"])
plt.figure(figsize=(10, 6))
sns.boxplot(x="amenity_group", y="occupancy_rate", data=analysis_data)
plt.title("Boxplot of Hotel Occupancy Rate by Amenity Quartile")
plt.xlabel("Number of Amenities Group")
plt.ylabel("Occupancy Rate")
plt.show()

print(occupancy_results)
print(review_results)

```

## Summary of Findings: Do Hotels with More Amenities Experience Higher Occupancy Rates or Review Scores?

1. Relationship Between Amenities and Hotel Performance

The regression models examine whether the number of amenities a hotel offers are related to its occupancy rate and review scores.
Two Ordinary Least Squares (OLS) regression models were created:
- Amenities vs. Occupancy Rate
- Amenities vs. Review Score

2. Key Regression Findings

- Amenities vs. Occupancy Rate
The regression output tells us how much occupancy rate changes when amenities increase. As we cann see, the coefficient for num_amenities is positive and statistically significant (p-value < 0.05), which suggests that having more amenities is linked to higher occupancy rates.

- Amenities vs. Review Score
Similar logic applies to review scores. As the coefficient for num_amenities is positive and staticstically significant, it suggests that hotels with more amenities tend to receive better reviews.

3. Visualization Interpretations

The left scatterplot shows the relationship between amenities and occupancy rate, whereas the right scatterplot shows the relationship between amenities and review scores. A clear upward trend in either plot suggests a positive relationship.

The boxplot presents the distribution of hotel occupancy rates across four different groups based on the number of amenities a hotel offers (Low, Medium-Low, Medium-High, High). Each box represents the interquartile range (IQR), which contains the middle 50% of the data. The horizontal line inside each box indicates the median occupancy rate for that category. The whiskers extend to show variability outside the IQR, and individual points beyond the whiskers represent outliers. By comparing the boxes, we can observe how occupancy rates vary with different levels of amenities. A noticeable upward trend in median values across groups suggests that hotels with more amenities tend to have higher occupancy rates. The boxplot helps verify whether higher-amenity hotels consistently have better occupancy (and as we can see, they do).
